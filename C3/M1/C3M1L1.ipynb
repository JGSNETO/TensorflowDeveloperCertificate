# Natural Language Processing in TensorFlow

## 1. Introduction to Natural Language Processing

**Natural Language Processing (NLP)** is the area of machine learning that deals with understanding and processing human language.

Examples of NLP applications include:

* Sentiment analysis
* Spam detection
* Text classification
* Machine translation
* Text generation
* Question answering
* Chatbots
* Search engines
* Speech recognition

The fundamental problem is that **neural networks work with numbers, not words**.

Therefore, we need to convert human language into a numerical representation that a neural network can process.

The general process is:

```text
Raw Text
   ↓
Tokenization
   ↓
Numerical Encoding
   ↓
Embedding
   ↓
Neural Network
   ↓
Prediction
```

---

# 2. Why Text Is Different from Images

Images are relatively straightforward for neural networks because they can naturally be represented as tensors.

For example, a grayscale image can be represented as pixel intensity values:

```text
[
    [0,   50,  255],
    [20,  100, 200],
    [10,  150, 250]
]
```

Each pixel has a numerical value.

Therefore:

```text
Image
  ↓
Pixels
  ↓
Numbers
  ↓
Tensor
  ↓
Neural Network
```

Text is much more complicated.

Consider:

```text
"The car is fast"
```

A neural network cannot directly process the words `"The"`, `"car"`, `"is"`, and `"fast"`.

We need to convert them into numbers.

---

# 3. Character-Based Encoding

One possible solution is to represent text using individual characters.

For example:

```text
"I love cars"
```

can be broken into:

```text
I
(space)
l
o
v
e
(space)
c
a
r
s
```

Each character can then be assigned a numerical value.

For example:

```text
I → 10
l → 11
o → 12
v → 13
e → 14
c → 15
a → 16
r → 17
s → 18
```

The sentence could then become something like:

```text
[10, 0, 11, 12, 13, 14, 0, 15, 16, 17, 18]
```

This approach is possible, but it has a major problem.

## The problem with character encoding

Character values don't contain the **meaning of the word**.

For example:

```text
car
vehicle
automobile
```

These words have similar meanings.

However, their individual characters and character encodings don't naturally tell the neural network that they are related.

The model would have to learn the relationship between individual characters and eventually discover the concept of a word.

This can be difficult.

---

# 4. Word-Based Encoding

Instead of treating individual characters as the basic unit, we can treat **words as tokens**.

Consider:

```text
"I love cars"
```

We can split this sentence into:

```text
["I", "love", "cars"]
```

Each word can then be assigned a unique numerical value:

```text
I     → 1
love  → 2
cars  → 3
```

The sentence becomes:

```text
[1, 2, 3]
```

This is called **word-based encoding**.

The important idea is:

> Each word receives a unique numerical ID that remains consistent throughout the dataset.

For example, if:

```text
car → 37
```

then every occurrence of `"car"` should be represented by the same ID.

---

# 5. The Numerical Value Does Not Represent Meaning

An important concept is that the numerical ID itself has **no semantic meaning**.

Suppose:

```text
car     → 1
truck   → 2
banana  → 3
```

It would be incorrect to conclude that:

```text
truck
```

is somehow more similar to:

```text
car
```

because:

```text
|2 - 1| = 1
```

Similarly, it would be incorrect to say that:

```text
banana
```

is less related to `"car"` because:

```text
|3 - 1| = 2
```

These numbers are simply **identifiers**.

They tell the model:

> "This number corresponds to this particular token."

They do not tell the model:

> "This word has this particular meaning."

This distinction is extremely important.

---

# 6. Tokenization

The process of splitting text into smaller units is called **tokenization**.

For simple word-based NLP, the tokens are usually words.

For example:

```text
"I love autonomous cars"
```

becomes:

```text
["I", "love", "autonomous", "cars"]
```

Each element is a **token**.

The process can be visualized as:

```text
"I love autonomous cars"
          ↓
      Tokenization
          ↓
["I", "love", "autonomous", "cars"]
```

Depending on the NLP system, tokens don't necessarily have to be complete words. They can also be:

* Characters
* Subwords
* Words
* Special tokens

Modern NLP systems often use **subword tokenization**, which helps deal with unknown or uncommon words.

---

# 7. Vocabulary

The collection of tokens that the tokenizer knows about is called the **vocabulary**.

Suppose the training data contains:

```text
"I love cars"
"I love autonomous cars"
"Cars are amazing"
```

The tokenizer might construct a vocabulary such as:

```text
cars        → 1
love        → 2
I           → 3
autonomous  → 4
are         → 5
amazing     → 6
```

The exact numerical assignments don't matter.

What matters is that the mapping is consistent.

For example:

```text
cars → 1
```

should always represent the same token.

---

# 8. Out-of-Vocabulary Words

What happens if the model encounters a word that wasn't present in the training vocabulary?

Suppose the vocabulary contains:

```text
car
truck
bus
vehicle
```

but the model encounters:

```text
"I bought a motorcycle"
```

The word:

```text
motorcycle
```

is unknown to the tokenizer.

This is called an **out-of-vocabulary (OOV)** word.

A special token can be used to represent unknown words:

```text
<OOV>
```

So the sentence could become:

```text
["I", "bought", "a", "<OOV>"]
```

This allows the model to continue processing the sentence even though it doesn't recognize every word.

---

# 9. The Problem with Integer Encoding

At this point, we have converted text into numbers:

```text
"I love cars"
      ↓
[1, 2, 3]
```

But there is still a problem.

The numbers themselves don't represent semantic relationships.

For example:

```text
car     → 1
truck   → 2
banana  → 3
```

There is no meaningful relationship between the numerical distances.

We therefore need a better representation.

This leads to one of the most important concepts in NLP:

# Word Embeddings

---

# 10. Word Embeddings

A **word embedding** represents a word using a vector of numerical values.

Instead of:

```text
car → 37
```

we can represent it as:

```text
car → [0.21, -0.73, 0.14, 0.62]
```

Another word might be:

```text
truck → [0.25, -0.69, 0.17, 0.58]
```

And another:

```text
banana → [-0.82, 0.31, 0.91, -0.44]
```

The embedding gives each word a position in a multidimensional numerical space.

---

# 11. Embeddings and Semantic Similarity

The important property of embeddings is that words with related meanings can have similar vector representations.

For example:

```text
car
truck
bus
vehicle
```

may have vectors that are relatively close together.

While:

```text
banana
apple
orange
```

may form another group.

Conceptually:

```text
              Animals
                 ↑
                 |
        dog      |      cat
                 |
                 |
─────────────────┼─────────────────→
                 |
       car       |      truck
                 |
                 |
              Vehicles
```

The actual embedding space is usually much larger than two dimensions.

It might contain:

```text
8 dimensions
32 dimensions
64 dimensions
128 dimensions
256 dimensions
...
```

The dimensions don't necessarily correspond to human-understandable concepts.

Instead, the neural network learns a useful representation from the training data.

---

# 12. Integer Encoding vs. Embedding

It is important to distinguish these two concepts.

### Integer encoding

Answers:

> **Which token is this?**

For example:

```text
car → 37
```

The number is an identifier.

### Embedding

Answers:

> **How can this token be represented numerically in a way that is useful for learning?**

For example:

```text
car → [0.21, -0.73, 0.14, 0.62, ...]
```

So the process is:

```text
Word
 ↓
Token ID
 ↓
Embedding vector
```

---

# 13. From Words to Tensors

Suppose we have:

```text
"I love cars"
```

which becomes:

```text
[1, 2, 3]
```

An embedding layer can transform these IDs into vectors:

```text
1 → [0.2, 0.7, -0.1]
2 → [0.8, 0.1,  0.4]
3 → [0.5, 0.9, -0.3]
```

Therefore:

```text
[1, 2, 3]
```

becomes:

```text
[
    [0.2, 0.7, -0.1],
    [0.8, 0.1,  0.4],
    [0.5, 0.9, -0.3]
]
```

Now the text has been transformed into a numerical tensor that can be processed by a neural network.

The complete process is:

```text
Raw text
   ↓
Tokenization
   ↓
Integer IDs
   ↓
Embedding
   ↓
Vectors
   ↓
Tensor
   ↓
Neural Network
```

---

# 14. The Problem of Different Sentence Lengths

There is another problem.

Consider:

```text
"I love cars"
```

which might become:

```text
[1, 2, 3]
```

But:

```text
"I really love autonomous cars"
```

might become:

```text
[1, 4, 2, 5, 3]
```

The sequences have different lengths.

This can make it difficult to process multiple sentences together as a batch.

We therefore commonly use **padding**.

---

# 15. Padding

Padding adds special values, usually zeros, to make sequences the same length.

For example, if the maximum length is 5:

```text
[1, 2, 3]
```

can become:

```text
[0, 0, 1, 2, 3]
```

while:

```text
[1, 4, 2, 5, 3]
```

already has length 5.

Now they can be represented together:

```text
[
    [0, 0, 1, 2, 3],
    [1, 4, 2, 5, 3]
]
```

Padding allows the data to have a consistent shape.

---

# 16. Why Word Order Matters

Another important challenge in NLP is that **word order matters**.

Consider:

```text
"The dog chased the cat."
```

and:

```text
"The cat chased the dog."
```

Both sentences contain the same words.

However, their meanings are different.

Therefore, NLP models need to learn not only:

> Which words are present?

but also:

> How are the words related to each other?

This is why NLP is fundamentally a **sequence-processing problem**.

---

# 17. Context Matters

Consider these two sentences:

```text
"The car is good."
```

and:

```text
"The car is not good."
```

The second sentence contains almost all the same words, but its meaning is very different.

The word:

```text
"not"
```

changes the meaning of the sentence.

Therefore, a good NLP model needs to consider **context**.

This is one reason why sequence-based neural networks became important in NLP.

---

# 18. Sequence Models

Different neural network architectures can be used to process sequences of text.

## RNN — Recurrent Neural Network

An RNN processes a sequence while maintaining information from previous elements.

Conceptually:

```text
Word 1
  ↓
Word 2
  ↓
Word 3
  ↓
Word 4
```

The model maintains some form of internal state as it processes the sequence.

This allows information from earlier words to influence later predictions.

---

# 19. LSTM — Long Short-Term Memory

An **LSTM** is a type of RNN designed to handle longer-term dependencies more effectively.

It uses internal mechanisms called **gates** to control what information should:

* Be remembered
* Be forgotten
* Be passed forward

Conceptually:

```text
Word
 ↓
LSTM
 ↓
Memory
 ↓
Next word
 ↓
LSTM
 ↓
Updated memory
```

LSTMs became widely used for tasks such as:

* Text classification
* Language modeling
* Translation
* Speech processing

---

# 20. GRU — Gated Recurrent Unit

A **GRU** is another type of recurrent neural network.

It has a similar goal to an LSTM but uses a somewhat simpler architecture.

Both LSTM and GRU are designed to help neural networks maintain useful information across sequences.

---

# 21. CNNs for Text

CNNs are not only useful for images.

They can also be applied to text using **1D convolutions**.

For example:

```text
"I really love this car"
```

A convolution could examine local groups of words:

```text
"I really love"
"really love this"
"love this car"
```

The model can learn useful local patterns.

For example:

```text
"very good"
"really good"
"absolutely fantastic"
```

could become patterns associated with positive sentiment.

Therefore:

```text
Image CNN
    ↓
Spatial patterns in images

Text CNN
    ↓
Local patterns in sequences
```

# 22. Complete NLP Classification Pipeline

Consider a movie review classification problem.

The training data might contain:

```text
"This movie was fantastic" → Positive
"This movie was terrible"  → Negative
"I really enjoyed it"      → Positive
"I hated this film"        → Negative
```

The overall pipeline becomes:

```text
                 RAW TEXT
                     ↓
                Tokenization
                     ↓
              Integer Encoding
                     ↓
                  Padding
                     ↓
                 Embedding
                     ↓
          Sequence / CNN / RNN
                     ↓
                Dense Layer
                     ↓
                 Prediction
                     ↓
          Positive / Negative
```

The neural network learns patterns in the numerical representation of the text and uses those patterns to make predictions.

# 26. Key Concepts to Remember

### Tokenization

Breaking text into smaller units called **tokens**.

```text
"I love cars"
      ↓
["I", "love", "cars"]
```

### Vocabulary

The collection of tokens known by the tokenizer.

```text
car → 37
truck → 52
bus → 81
```

### Integer Encoding

Converting tokens into numerical IDs.

```text
["I", "love", "cars"]
          ↓
[1, 2, 3]
```

### OOV

A special representation for words that are not in the vocabulary.

```text
motorcycle → <OOV>
```

### Padding

Adding values to sequences so they have the same length.

```text
[1, 2, 3]
     ↓
[0, 0, 1, 2, 3]
```

### Embedding

Representing tokens as numerical vectors that can capture useful relationships between words.

```text
car → [0.21, -0.73, 0.14, ...]
```

### Sequence Model

A model capable of processing information where **order and context matter**.

Examples:

* RNN
* LSTM
* GRU

# 27. The Most Important Distinction

Keep this distinction in mind as you continue the TensorFlow course:

```text
TOKEN ID
   ↓
"Which word is this?"
```

versus:

```text
EMBEDDING
   ↓
"How can this word be represented numerically
so the neural network can learn useful relationships?"
```

For example:

```text
car
 ↓
Token ID
 ↓
37
 ↓
Embedding
 ↓
[0.21, -0.73, 0.14, 0.62, ...]
```

The **37 does not represent the meaning of "car"**.

The embedding is the learned representation that allows the neural network to discover useful relationships between words.

# Using API

````python
import tensorflow as tf

sentences = [
    'I love my dog',
    'I love my cat'
]

vectorize_layer = tf.keras.layers.TextVectorization()
vectorize_layer.adapt(sentences)
vocabulary = vectorize_layer.get_vocabulary(include_special_tikens = False)
print(vocabulary)
````

````python
import tensorflow as tf

sentences = [
    'I love my dog',
    'I love my cat',
    'you love my dog!'
]

vectorize_layer = tf.keras.layers.TextVectorization()
vectorize_layer.adapt(sentences)
vocabulary = vectorize_layer.get_vocabulary()
print(vocabulary)
````
