# Machine Learning and Deep Learning — Core Definitions

## 1. Artificial Intelligence, Machine Learning, Deep Learning, and Reinforcement Learning

These concepts are related, but they describe different levels or paradigms of artificial intelligence.

```text
Artificial Intelligence (AI)
│
├── Machine Learning (ML)
│   │
│   ├── Supervised Learning
│   ├── Unsupervised Learning
│   └── Semi-Supervised Learning
│
├── Deep Learning (DL)
│   │
│   ├── Feedforward Neural Networks
│   ├── CNN
│   ├── RNN
│   │   ├── LSTM
│   │   └── GRU
│   └── Transformers
│
└── Reinforcement Learning (RL)
    │
    └── Deep Reinforcement Learning
```

---

# 2. Machine Learning (ML)

## Definition

**Machine Learning** is a field of Artificial Intelligence in which algorithms learn patterns from data and use those patterns to make predictions, classifications, or decisions.

Instead of explicitly programming every rule, we provide data and allow the model to learn a relationship between inputs and outputs.

For example, instead of writing:

```text
IF house_size > 200 m²
AND bedrooms > 3
AND location = Stockholm
THEN price = ...
```

we provide many examples:

```text
House features → Actual price
--------------------------------
120 m²         → 3.2 M SEK
180 m²         → 4.5 M SEK
250 m²         → 6.8 M SEK
...
```

The model learns a relationship:

```text
Features → Model → Prediction
```

---

# 3. Types of Machine Learning

## 3.1 Supervised Learning

In **supervised learning**, the training data contains both:

* **Features** → input
* **Labels** → expected output

```text
Features ───────► Model ───────► Prediction
   │                               │
   └──────── Label ◄───────────────┘
```

The model compares its prediction with the known label and adjusts itself during training.

### Common supervised learning tasks

### Classification

Predict a discrete category.

Examples:

```text
Email → Spam / Not Spam
Image → Cat / Dog
Vehicle → Safe / Unsafe
Transaction → Fraud / Not Fraud
```

### Regression

Predict a continuous numerical value.

Examples:

```text
Vehicle speed → 82.5 km/h
House → 4.8 million SEK
Temperature → 21.7 °C
Battery SOC → 73.2 %
```

---

# 4. Unsupervised Learning

In **unsupervised learning**, the training data does not contain labels.

The model attempts to discover patterns or structures within the data.

Examples include:

* Clustering
* Dimensionality reduction
* Anomaly detection
* Association analysis

Example:

```text
Customer data
     ↓
Clustering algorithm
     ↓
Group 1
Group 2
Group 3
```

The algorithm determines the groups rather than being explicitly told what each group represents.

---

# 5. Semi-Supervised Learning

**Semi-supervised learning** combines:

* A relatively small amount of labeled data
* A larger amount of unlabeled data

Example:

```text
10,000 images
│
├── 500 labeled images
└── 9,500 unlabeled images
```

This can be useful when obtaining labels is expensive or time-consuming.

---

# 6. Reinforcement Learning (RL)

## Definition

**Reinforcement Learning** is a machine learning paradigm where an **agent interacts with an environment**, takes actions, and receives rewards or penalties.

```text
             Action
Agent ─────────────────► Environment
  ▲                           │
  │                           │
  └──── State + Reward ◄──────┘
```

The objective is to learn a **policy** that maximizes cumulative reward.

### Example: Autonomous vehicle

```text
Environment
    ↓
Vehicle observes state
    ↓
RL Agent
    ↓
Action
    ├── Accelerate
    ├── Brake
    ├── Steer left
    └── Steer right
    ↓
Environment
    ↓
Reward
```

A reward could represent:

```text
+10 → reaches target
+1  → maintains lane
-5  → unnecessary braking
-100 → collision
```

---

# 7. Deep Learning (DL)

## Definition

**Deep Learning** is a subset of machine learning that uses neural networks containing multiple layers to learn increasingly complex representations of data.

```text
Input
  ↓
Layer 1
  ↓
Layer 2
  ↓
Layer 3
  ↓
...
  ↓
Output
```

Deep learning is particularly useful for complex and high-dimensional data such as:

* Images
* Video
* Audio
* Natural language
* Sensor data
* Time-series data

---

# 8. Traditional ML vs Deep Learning

A simplified comparison:

| Traditional ML                         | Deep Learning                           |
| -------------------------------------- | --------------------------------------- |
| Often requires feature engineering     | Can learn features automatically        |
| Often works well with smaller datasets | Often benefits from larger datasets     |
| Models can be relatively simple        | Neural networks can contain many layers |
| CPU often sufficient                   | GPUs/TPUs are commonly used             |
| Examples: SVM, Random Forest           | CNN, RNN, LSTM, Transformer             |

Feature engineering is not completely eliminated by deep learning, but neural networks can learn useful representations directly from raw or minimally processed inputs.

# 9. Features

## Definition

**Features** are the input variables used by a machine learning model.

For a vehicle:

```text
Features:

speed
steering_angle
acceleration
yaw_rate
brake_pressure
vehicle_mass
```

These values are provided to the model:

```text
Features
   ↓
Model
   ↓
Prediction
```

Features can be:

### Numerical

```text
speed = 72.5
temperature = 21.4
mass = 1800
```

### Categorical

```text
fuel_type = "electric"
vehicle_type = "SUV"
```

Categorical data generally needs to be encoded before being used by a neural network.

---

# 10. Labels

## Definition

A **label** is the target value that a supervised learning model attempts to predict.

Example:

```text
Features:
speed = 72 km/h
steering = 3°
rain = true

Label:
braking_required = true
```

For regression:

```text
Features → Vehicle sensors
Label    → Stopping distance = 38.5 m
```

For classification:

```text
Features → Vehicle sensors
Label    → Collision / No Collision
```

---

# 11. Neural Networks

A neural network consists of interconnected computational units called **neurons**.

A simple network:

```text
Input Layer       Hidden Layer       Output Layer

 x₁ ───────────►    ○
                   │
 x₂ ───────────►   ○ ───────────► Prediction
                   │
 x₃ ───────────►   ○
```

Each neuron performs a mathematical operation based on its inputs, weights, bias, and activation function.

A simplified neuron is:

```text
z = w₁x₁ + w₂x₂ + ... + wₙxₙ + b
```

The activation function is then applied:

```text
output = activation(z)
```

---

# 12. Weights and Bias

## Weights

Weights determine how strongly each input contributes to a neuron's output.

```text
x₁ ── w₁ ──┐
x₂ ── w₂ ──┼──► Neuron
x₃ ── w₃ ──┘
```

During training, the model adjusts the weights to reduce the prediction error.

## Bias

The bias provides an additional trainable parameter:

```text
z = Wx + b
```

Both weights and biases are learned during training.

---

# 13. Activation Functions

An **activation function** transforms the output of a neuron and, importantly, introduces **non-linearity** into the network.

Without nonlinear activation functions, stacking multiple linear layers would still result in an overall linear transformation.

Therefore:

```text
Linear layers only
       ↓
Still fundamentally linear

Linear + Nonlinear activation
       ↓
Can model complex nonlinear relationships
```

---

# 14. Common Activation Functions

## 14.1 ReLU

The **Rectified Linear Unit** is:

```text
ReLU(x) = max(0, x)
```

Examples:

```text
x = -5 → 0
x =  0 → 0
x =  3 → 3
```

ReLU is widely used in hidden layers of neural networks.

TensorFlow/Keras:

```python
keras.layers.Dense(64, activation="relu")
```

---

## 14.2 Sigmoid

Sigmoid maps values approximately into:

```text
0 → 1
```

Formula:

```text
σ(x) = 1 / (1 + e⁻ˣ)
```

It is commonly used for the output of a **binary classification** model.

Example:

```python
keras.layers.Dense(1, activation="sigmoid")
```

Output:

```text
0.87
```

This can be interpreted as a model probability estimate for the positive class, assuming the model and labels are configured accordingly.

---

## 14.3 Softmax

Softmax converts multiple outputs into values that sum to approximately 1.

Example:

```text
Cat       → 0.10
Dog       → 0.80
Horse     → 0.10
```

Commonly used for multi-class classification.

```python
keras.layers.Dense(3, activation="softmax")
```

---

## 14.4 Tanh

Tanh produces values between:

```text
-1 and +1
```

It has historically been common in recurrent neural networks and can still be useful in specific architectures.

---

# 15. Choosing Activation Functions

| Situation                         | Common choice              |
| --------------------------------- | -------------------------- |
| Hidden layers                     | ReLU / variants            |
| Binary classification output      | Sigmoid                    |
| Multi-class classification output | Softmax                    |
| Regression output                 | Often linear/no activation |
| Some recurrent architectures      | Tanh / Sigmoid             |

The activation function should be chosen based on the role of the layer and the mathematical nature of the output.

---

# 16. Loss Function

A **loss function** measures how different the model's prediction is from the expected target.

```text
Prediction
     │
     ▼
Loss Function ◄──── True Label
     │
     ▼
   Loss
```

Example:

```text
True value       = 100
Predicted value  = 92

Error = 8
```

The exact loss depends on the task.

Common examples:

| Task                       | Common loss               |
| -------------------------- | ------------------------- |
| Binary classification      | Binary Cross-Entropy      |
| Multi-class classification | Categorical Cross-Entropy |
| Regression                 | Mean Squared Error (MSE)  |
| Regression                 | Mean Absolute Error (MAE) |

---

# 17. Optimizer

The optimizer updates the model's trainable parameters to reduce the loss.

A simplified training loop:

```text
Input
  ↓
Prediction
  ↓
Loss
  ↓
Gradients
  ↓
Optimizer
  ↓
Update weights
  ↓
Repeat
```

Common optimizers include:

* SGD
* Adam
* RMSprop

Example:

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

# 18. Convolutional Neural Networks (CNNs)

## Definition

A **Convolutional Neural Network (CNN)** is a neural network architecture designed to efficiently process data with spatial or local structure.

CNNs are particularly associated with:

* Images
* Video
* Computer vision
* Spatial sensor data

They can also be applied to some time-series and other structured signals.

---

# 19. How a CNN Works

Instead of connecting every input pixel to every neuron immediately, a CNN applies **filters/kernels** that move across the input.

Simplified:

```text
Input Image

┌───────────────┐
│ pixels        │
│               │
│               │
└───────────────┘
        ↓
   Convolution
        ↓
 Feature Map
        ↓
   Activation
        ↓
    Pooling
        ↓
    More CNN
     Layers
        ↓
   Classification
```

The filters learn to detect useful local patterns.

Early layers might learn representations corresponding to:

```text
Edges
 ↓
Corners
 ↓
Textures
 ↓
Shapes
 ↓
Objects
```

---

# 20. Convolution

A convolution applies a small matrix called a **kernel** across the input.

Conceptually:

```text
Input

[ 1  2  3 ]
[ 4  5  6 ]
[ 7  8  9 ]

      ↓

Kernel

[ 1  0 ]
[ 0 -1 ]
```

The kernel moves across the image and produces a **feature map**.

During training, the network learns the values of these filters.

---

# 21. CNN Layers

A typical CNN might look like:

```text
Input
  ↓
Conv2D
  ↓
ReLU
  ↓
MaxPooling
  ↓
Conv2D
  ↓
ReLU
  ↓
MaxPooling
  ↓
Flatten
  ↓
Dense
  ↓
Output
```

TensorFlow/Keras example:

```python
model = keras.Sequential([
    keras.layers.Conv2D(
        32,
        kernel_size=3,
        activation="relu",
        input_shape=(64, 64, 3)
    ),
    keras.layers.MaxPooling2D(),

    keras.layers.Conv2D(
        64,
        kernel_size=3,
        activation="relu"
    ),
    keras.layers.MaxPooling2D(),

    keras.layers.Flatten(),

    keras.layers.Dense(
        64,
        activation="relu"
    ),

    keras.layers.Dense(
        10,
        activation="softmax"
    )
])
```

---

# 22. Pooling

Pooling reduces the spatial dimensions of feature maps.

A common method is **Max Pooling**.

Example:

```text
[1  3]
[2  4]

Max = 4
```

Max pooling helps:

* Reduce computational cost
* Reduce spatial dimensions
* Increase the effective receptive field
* Provide some robustness to small spatial changes

# 23. Recurrent Neural Networks (RNNs)

## Definition

A **Recurrent Neural Network (RNN)** is designed to process sequential data.

Examples:

* Text
* Speech
* Time series
* Sensor signals
* Vehicle telemetry

The important characteristic is that the network maintains information from previous time steps.

```text
x₁ → RNN → h₁
          ↓
x₂ → RNN → h₂
          ↓
x₃ → RNN → h₃
          ↓
x₄ → RNN → h₄
```

The hidden state carries information forward through the sequence.

---

# 24. RNN Concept

For each time step:

```text
Current Input
     │
     ▼
   ┌─────┐
──►│ RNN │──► Output
   └─────┘
      ▲
      │
Previous Hidden State
```

Conceptually:

```text
hₜ = f(xₜ, hₜ₋₁)
```

The current hidden state depends on:

* Current input `xₜ`
* Previous hidden state `hₜ₋₁`

This gives RNNs their ability to process sequential information.

---

# 25. Problems with Basic RNNs

Basic RNNs can have difficulty learning long-term dependencies because of problems such as:

* Vanishing gradients
* Exploding gradients

For example:

```text
"The car stopped because the traffic light was red."
```

The model may need information from earlier parts of the sequence to interpret later information.

As sequences become longer, maintaining useful information across many recurrent steps can become difficult.

---

# 26. LSTM

**Long Short-Term Memory (LSTM)** is a type of recurrent neural network designed to better preserve information over longer sequences.

An LSTM contains gates that control information flow.

Conceptually:

```text
             ┌──────────────┐
Input ──────►│              │
             │     LSTM     │──► Output
Previous ───►│              │
State        └──────────────┘
```

The major components include:

* Forget gate
* Input gate
* Output gate
* Cell state

These mechanisms allow the network to learn what information should be:

```text
Forgotten
Stored
Updated
Output
```

---

# 27. GRU

**Gated Recurrent Unit (GRU)** is another recurrent architecture.

GRUs provide gating mechanisms similar in purpose to LSTMs but use a simpler structure.

```text
RNN
├── Basic RNN
├── LSTM
└── GRU
```

Both LSTM and GRU can be useful for sequential data.

---

# 28. CNN vs RNN

| CNN                                        | RNN                                     |
| ------------------------------------------ | --------------------------------------- |
| Primarily processes spatial/local patterns | Designed for sequential relationships   |
| Common in computer vision                  | Common in sequence processing           |
| Uses convolutional filters                 | Uses recurrent hidden states            |
| Images, video, spatial signals             | Text, time series, sensor sequences     |
| Processes local receptive fields           | Processes information across time steps |

They can also be combined.

For example:

```text
Video
 ↓
CNN
 ↓
Visual features per frame
 ↓
LSTM
 ↓
Temporal understanding
 ↓
Prediction
```

---

# 29. Transformers

Transformers are neural network architectures designed to process sequences using **attention mechanisms**.

They have become highly important in:

* NLP
* Large Language Models
* Computer vision
* Multimodal AI
* Time-series modeling

Instead of processing tokens strictly one at a time like a basic RNN, attention allows the model to consider relationships between different positions in a sequence.

Simplified:

```text
Input sequence
      ↓
Embeddings
      ↓
Positional information
      ↓
Self-Attention
      ↓
Feedforward layers
      ↓
More Transformer blocks
      ↓
Output
```

---

# 30. CNN, RNN, and Transformer Relationship

A useful way to remember the architectures:

```text
Neural Networks
│
├── Feedforward Networks
│
├── CNN
│   └── Spatial / local patterns
│
├── RNN
│   ├── Basic RNN
│   ├── LSTM
│   └── GRU
│       └── Sequential / temporal patterns
│
└── Transformer
    └── Attention / relationships between positions
```

The boundaries are not absolute. Modern architectures can combine several ideas.

# 31. Embeddings

An **embedding** converts discrete values such as words or tokens into dense numerical vectors.

For example:

```text
"car" → [0.21, -0.43, 0.78, ...]
```

Instead of representing a word only as an integer:

```text
car → 127
```

an embedding represents it as a vector containing learned numerical features.

In TensorFlow/Keras:

```python
embedding = keras.layers.Embedding(
    input_dim=10000,
    output_dim=128
)
```

Meaning:

```text
Vocabulary size = 10,000
Embedding dimension = 128
```

---

# 32. Padding and Masking

When working with NLP, sequences often have different lengths.

Example:

```text
[12, 45, 89]

[12, 45, 76, 89]

[12, 34, 45, 76, 91, 89]
```

Padding makes them the same length:

```text
[12, 45, 89,  0,  0,  0]

[12, 45, 76, 89,  0,  0]

[12, 34, 45, 76, 91, 89]
```

The `0` values represent padding.

TensorFlow:

```python
padded = keras.utils.pad_sequences(
    sequences,
    padding="post"
)
```

When appropriate, an embedding layer can use:

```python
keras.layers.Embedding(
    input_dim=10000,
    output_dim=128,
    mask_zero=True
)
```

The mask tells compatible downstream layers which positions are padding.

# 33. Training a Neural Network

A simplified training process:

```text
Dataset
   ↓
Features + Labels
   ↓
Preprocessing
   ↓
Model
   ↓
Prediction
   ↓
Loss
   ↓
Backpropagation
   ↓
Gradients
   ↓
Optimizer
   ↓
Update weights
   ↓
Repeat
```

This process happens over many iterations.

# 34. Epochs, Batches, and Iterations

## Batch

A **batch** is a subset of the training data processed together.

Example:

```text
Dataset = 10,000 samples
Batch size = 32
```

The model processes 32 samples at a time.

## Epoch

One **epoch** means the training algorithm has processed the entire training dataset once.

```text
10,000 samples
      ↓
Multiple batches
      ↓
1 epoch
```

## Iteration / Step

One training step usually corresponds to processing one batch and updating the model parameters.

# 35. Training, Validation, and Test Data

A dataset is commonly divided into:

```text
Dataset
│
├── Training Data
│      ↓
│   Learn parameters
│
├── Validation Data
│      ↓
│   Tune / evaluate during development
│
└── Test Data
       ↓
    Final evaluation
```

The test set should generally be kept separate from model development decisions.

# 36. Overfitting and Underfitting

## Overfitting

The model performs very well on training data but poorly on unseen data.

```text
Training performance → Excellent
Validation performance → Poor
```

The model has learned the training data too specifically.

Common techniques to reduce overfitting include:

* More training data
* Data augmentation
* Regularization
* Dropout
* Early stopping
* Reducing model complexity

## Underfitting

The model is unable to capture the underlying patterns.

```text
Training performance → Poor
Validation performance → Poor
```

Possible solutions include:

* More expressive models
* Better features
* More training
* Reduced regularization

---

# 37. Generalization

The ultimate objective is not simply to memorize the training data.

The model should **generalize** to previously unseen examples.

```text
Training data
     ↓
     Model
     ↓
Unseen data
     ↓
Useful predictions
```

Good generalization is one of the central goals of machine learning.
