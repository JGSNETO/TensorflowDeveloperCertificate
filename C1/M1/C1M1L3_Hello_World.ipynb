# Keras Summary

**Keras** is a high-level deep learning API that makes it easy to build, train, evaluate, and deploy neural networks. It provides a simple, user-friendly interface while relying on a powerful backend (typically TensorFlow) to perform the underlying computations.

## Key Features

- Easy to use with a clean and intuitive API.
- Integrated into TensorFlow as **`tf.keras`**.
- Modular and extensible architecture.
- Supports rapid prototyping and production-ready applications.
- Can run on **CPU**, **GPU**, and **TPU**.

---

## Main Components

| Component | Purpose |
|-----------|---------|
| **Models** | Define the overall architecture of the neural network. |
| **Layers** | Building blocks of a model (e.g., `Dense`, `Conv2D`, `LSTM`). |
| **Activation Functions** | Introduce non-linearity (e.g., ReLU, Sigmoid, Softmax). |
| **Loss Functions** | Measure how well the model performs during training. |
| **Optimizers** | Update the model's weights to minimize the loss (e.g., Adam, SGD). |
| **Metrics** | Evaluate the model's performance (e.g., Accuracy, Precision). |

---

## Keras Workflow

Building a deep learning model in Keras typically follows five steps:

1. Build the model.
2. Compile the model.
3. Train the model.
4. Evaluate the model.
5. Make predictions.

## Common Layer Types

| Layer | Description | Typical Applications |
|--------|-------------|----------------------|
| `Dense` | Fully connected layer used in most neural networks. | Classification, regression, binary prediction, final output layers. |
| `Conv2D` | Convolutional layer for image processing. | Image classification, object detection, image segmentation, medical imaging. |
| `MaxPooling2D` | Reduces the spatial dimensions of feature maps. | CNNs for image processing to reduce computation and improve feature extraction. |
| `Flatten` | Converts multidimensional data into a one-dimensional vector. | Connecting convolutional layers to fully connected (`Dense`) layers. |
| `Dropout` | Reduces overfitting by randomly disabling neurons during training. | Regularization in deep neural networks to improve generalization. |
| `LSTM` | Long Short-Term Memory layer for sequential data. | Time series forecasting, speech recognition, language modeling, text generation. |
| `GRU` | Gated Recurrent Unit for sequence modeling. | Time series prediction, NLP, speech processing, sequential sensor data. |
| `Embedding` | Converts words or tokens into dense vectors for NLP tasks. | Text classification, sentiment analysis, machine translation, chatbots, language models. |

## Popular Optimizers

| Optimizer | Description | Best Used For |
|-----------|-------------|---------------|
| **Adam (Adaptive Moment Estimation)** | Combines Momentum and RMSprop by adapting the learning rate for each parameter using estimates of the first and second moments of the gradients. It is one of the most popular optimizers due to its fast convergence and ease of use. | General-purpose deep learning, CNNs, RNNs, NLP, computer vision, and most projects where you need a strong default optimizer. |
| **SGD (Stochastic Gradient Descent)** | Updates model weights using the gradient computed from a single training example (or a mini-batch). It is simple, memory-efficient, and often provides good generalization, especially when combined with Momentum. | Large datasets, image classification, and situations where better generalization is more important than fast convergence. |
| **RMSprop (Root Mean Square Propagation)** | Adjusts the learning rate for each parameter based on the recent magnitude of its gradients. This helps stabilize training and speeds up convergence for non-stationary problems. | Recurrent Neural Networks (RNNs), LSTMs, sequential data, and problems with noisy or changing gradients. |
| **Adagrad (Adaptive Gradient Algorithm)** | Adapts the learning rate individually for each parameter by giving larger updates to infrequent features and smaller updates to frequent ones. The learning rate continuously decreases during training. | Sparse datasets, Natural Language Processing (NLP), recommendation systems, and text classification. |
| **AdamW** | An improved version of Adam that separates (decouples) weight decay from gradient updates, leading to better regularization and improved generalization. It is widely used in modern deep learning models. | Transformer models, large neural networks, computer vision, NLP, and deep learning models requiring effective regularization. |

## Evaluation Metrics

| Metric | Description | Best Used For |
|--------|-------------|---------------|
| **Accuracy** | Measures the proportion of correct predictions out of all predictions made. | General classification problems where the classes are balanced. |
| **Precision** | Measures the proportion of predicted positive instances that are actually positive. It answers: *"Of everything the model predicted as positive, how many were correct?"* | Problems where false positives are costly, such as spam detection or fraud detection. |
| **Recall** | Measures the proportion of actual positive instances that were correctly identified. It answers: *"Of all the actual positives, how many did the model find?"* | Problems where false negatives are costly, such as disease diagnosis or safety-critical systems. |
| **F1-Score** | The harmonic mean of Precision and Recall. It balances both metrics into a single score, making it useful when classes are imbalanced. | Classification tasks with imbalanced datasets where both Precision and Recall are important. |
| **Mean Absolute Error (MAE)** | Measures the average absolute difference between predicted and actual values. Every error contributes proportionally to the final score. Lower values indicate better performance. | Regression problems where all prediction errors should be weighted equally. |
| **Mean Squared Error (MSE)** | Measures the average squared difference between predicted and actual values. Squaring penalizes larger errors more heavily than smaller ones. Lower values indicate better performance. | Regression problems where large prediction errors should be penalized more strongly. |
| **Root Mean Squared Error (RMSE)** | The square root of the Mean Squared Error. It has the same units as the target variable, making it easier to interpret than MSE while still emphasizing large errors. | Regression tasks where interpretability and sensitivity to large errors are both important. |

## Core Concepts

### Tensor

A **tensor** is the fundamental data structure in TensorFlow. It is a multidimensional array used to store data.

Examples:

- Scalar → 0D tensor
- Vector → 1D tensor
- Matrix → 2D tensor
- Higher-dimensional arrays → 3D, 4D, etc.

```python
import tensorflow as tf

scalar = tf.constant(5)
vector = tf.constant([1, 2, 3])
matrix = tf.constant([[1, 2], [3, 4]])
```

---

### Operations

TensorFlow performs computations through operations on tensors.

Examples include:

- Addition
- Multiplication
- Matrix multiplication
- Reshaping
- Reduction
- Mathematical functions

```python
a = tf.constant([1, 2, 3])
b = tf.constant([4, 5, 6])

c = tf.add(a, b)
```

---

### Variables

Variables store parameters that can be updated during training, such as neural network weights.

```python
weights = tf.Variable([1.0, 2.0])
```

---

### Automatic Differentiation

TensorFlow automatically computes gradients using **GradientTape**, making backpropagation straightforward.

```python
x = tf.Variable(3.0)

with tf.GradientTape() as tape:
    y = x ** 2

gradient = tape.gradient(y, x)
```

# Important functions

## Loss function:
- Measures how good and how bad the guess done by the algorithm was. Then gives the data to the optimizer function.

## Optimazer function:

- It will perform the next guess, based on the Loss function input.

## Fit command:

- Where the training takes place.

## Predict method:

- Gives back values based on the training.


````python
import tensorflow as tf
import numpy as np
from tensorflow import keras

print(tf.__version__)
````

````python
# Build a simple Sequential Model: 1 layer and 1 Neuron
model = tf.keras.Sequential([keras.layers.Dense(units=1, input_shape=[1])])
````

````python
# Compile the model
model.compile(optimizer='sgd', loss='mean_squared_error')
````

````python
# Declare model inputs and outputs for training
xs = np.array([-1.0,  0.0, 1.0, 2.0, 3.0, 4.0], dtype=float)
ys = np.array([-3.0, -1.0, 1.0, 3.0, 5.0, 7.0], dtype=float)
````

````python
# Training the model
model.fit(xs, ys, epochs=1000)
````

````python
# Make a prediction
print(model.predict([10.0]))
````
