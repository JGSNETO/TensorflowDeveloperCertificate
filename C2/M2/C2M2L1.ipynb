# Custom training loops in machine learning

- A custom training loop in Tensorflow allows you to control every aspect of the training process, from forward propagation to backpropagation and gradient updates. This approach is useful when you need flexibility beyond TensorFlow built-in .fit() method, such as handling multiple losses, implementing custom training logic or fine-tunning optimizers. 

## Key concepts in Custom Training Loops

- A training loop consists of the following essential steps:
1. Forward Pass: Compute prediction from input data;
2. Compute loss: Measure the difference between predictions and true values;
3. Compute gradients: Use automatic differentiation(tf.GradientTape()) to compute gradients;
4. Update Weights: Apply an optimization algorithm(e.g, SGD, Adam) to update model weights;
5. Repeat for epochs: Iterate through the dataset multiple times(Epoch) to train the model.

## Implementing a custom training loop

```python
# Step 1: Import dependencies

import tensorflow as tf
import numpy as np

# Set random seed for reproducibility
tf.random.set_seed(42)
np.random.seed(42)
```

```python
# Step 2: Define a simple model: y = Wx + b

class LinearModel(tf.keras.Model):
    def __init__(self):
        super().__init__()
        self.dense = tf.keras.layers.Dense(1, activation = None)
    
    def call(self, inputs):
        return self.dense(inputs)
    
# Instantiate model
model = LinearModel()
```

```python
# Step 3: Prepare Data

# Generate synthetic data: y = 3x + 2 + noise

X_train = np.random.rand(100, 1).astype(np.float32)
y_train = 3 * X_train + 2 + 0.1 * np.random.randn(100, 1).astype(np.float32)

# Convert to tensorflow dataset
dataset = tf.data.Dataset.from_tensor_slices((X_train, y_train)).batch(10) # Batch size 10
```

```python
# Step 4: Define loss function and optimizer

# Mean square error (MSE) Loss
loss_fn = tf.keras.losses.MeanSquareError()

# Stochastic Gradient Descent (SGD) Optimizer
optimizer = tf.keras.optimizers.SGD(learning_rate = 0.1)
```

```python
# Step 5: Implementing the training loop

# Number of training epochs
epochs = 100

# Training loop

for epoch in range(epochs):
    for x_batch, y_batch in dataset:
        with tf.GradientTape() as tape:
            y_pred = model(x_batch) # Forward pass
            loss = loss_fn(y_batch, y_pred) # Compute loss

        # Compute gradients
        gradients = tape.gradients(loss, model.trainable_variables)

        # Apply gradients(Update weights)
        optimizer.apply_gradients(zip(gradients, model.trainable_variables))

    # Print progress every 10 epochs
    if epoch % 10 == 0:
        print(f"Epoch {epoch}, Loss: {loss.numpy():.4f}")
```

# Training Loop

- Start -> Initialize model <For each epoch> -> Calculate loss for input sample -> Calculate gradients of loss using gradient tape -> Update model trainable weights using gradients and learning rate. 

# Explanation of key steps
1. tf.GradientTape()
- TensorFlow´s automatic differentiation system;
- Records operations for computing gradients

2. loss_fn(y_batch, y_pred):
- Computes how far predictions are from actual values using MSE;

3. tape.gradient(loss, model.trainable_variables):
- Computes gradients of the loss with respect to model parameters;

4. optimizer.apply_gradients()
- Uses stochastic Gradient Descent(SGD) to update weight

# When to use custom training loops ?

- Need full control over training behavior;
- Implement complex loss functions or training schedules;
- Fine-tune gradients(e.g: gradient clipping);
- Work with multiple models or datasets.
