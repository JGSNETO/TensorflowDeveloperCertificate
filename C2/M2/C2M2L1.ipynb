# Custom Training Loops in Machine Learning

A **custom training loop** in TensorFlow gives you direct control over the training process of a neural network.

Instead of relying entirely on TensorFlow's built-in `.fit()` method, you explicitly define how the model:

1. Receives input data
2. Performs the forward pass
3. Calculates predictions
4. Computes the loss
5. Calculates gradients
6. Updates its weights
7. Repeats the process across batches and epochs

This approach is useful when the standard `.fit()` workflow is not flexible enough for the problem being solved.

---

# 1. Standard Training with `.fit()`

In many TensorFlow projects, the training process can be handled using the built-in `.fit()` method.

Conceptually:

```text
Training Data
      ↓
    Model
      ↓
    Loss
      ↓
  Optimizer
      ↓
 Updated Weights
      ↓
    Repeat
```

TensorFlow handles most of the details automatically.

This is convenient and is usually the preferred approach when the training process follows a standard pattern.

However, some machine learning problems require more control.

For example, you may want to:

* Use multiple loss functions
* Apply different optimization rules to different parts of a model
* Implement custom gradient calculations
* Apply special weight-update rules
* Train multiple models together
* Implement reinforcement-learning algorithms
* Dynamically change the training behavior
* Implement research-oriented or experimental architectures
* Fine-tune how gradients are calculated and applied

In these situations, a **custom training loop** can provide the required flexibility.

---

# 2. What Is a Custom Training Loop?

A custom training loop is a manually controlled training process in which you explicitly define the major steps involved in training a neural network.

The basic process is:

```text
              Training Data
                    ↓
              Forward Pass
                    ↓
               Prediction
                    ↓
              Compute Loss
                    ↓
            Compute Gradients
                    ↓
            Update Weights
                    ↓
              Next Batch
                    ↓
              Next Epoch
                    ↓
                  Repeat
```

The important difference is that **you control these steps instead of allowing `.fit()` to manage them automatically**.

---

# 3. Key Concepts in Custom Training Loops

A typical custom training loop consists of five major steps:

1. **Forward Pass**
2. **Compute Loss**
3. **Compute Gradients**
4. **Update Weights**
5. **Repeat for Multiple Epochs**

---

# 4. Step 1 — Forward Pass

The **forward pass** is the process of sending input data through the neural network to generate predictions.

Conceptually:

```text
Input
  ↓
Layer 1
  ↓
Layer 2
  ↓
Layer 3
  ↓
Output
  ↓
Prediction
```

For example, suppose a model receives an image.

The image is passed through the neural network:

```text
Image
  ↓
Convolution
  ↓
Activation
  ↓
Pooling
  ↓
Dense Layer
  ↓
Prediction
```

The output of the model represents its prediction.

For a classification problem, the output might represent the probability that an image belongs to a particular class.

---

# 5. Step 2 — Compute the Loss

After obtaining a prediction, we need to determine how accurate that prediction was.

This is done using a **loss function**.

The loss function compares:

```text
Prediction
     ↓
     vs.
     ↓
True Label
```

and produces a numerical value representing the model's error.

For example:

```text
True value:       1
Model prediction: 0.8
Loss:             relatively small
```

Another prediction might be:

```text
True value:       1
Model prediction: 0.1
Loss:             relatively large
```

The goal of training is to minimize the loss.

Therefore:

> **Loss measures how wrong the model is.**

---

# 6. Step 3 — Compute Gradients

Once the loss has been calculated, we need to determine how the model's weights should change to reduce that loss.

This is where **gradients** become important.

A gradient tells us how the loss changes with respect to a model parameter.

Conceptually:

```text
Loss
 ↓
How does the loss change if this weight changes?
 ↓
Gradient
```

For a particular weight:

$$
\frac{\partial L}{\partial w}
$$

where:

* \(L\) = loss
* \(w\) = model weight
* \(\frac{\partial L}{\partial w}\) = gradient of the loss with respect to that weight

TensorFlow provides **automatic differentiation** through `tf.GradientTape()`.

The conceptual process is:

```text
Forward Pass
     ↓
Calculate Loss
     ↓
GradientTape
     ↓
Calculate Gradients
     ↓
Gradients for Model Parameters
```

This is one of the most important components of a custom training loop.

---

# 7. Automatic Differentiation

Calculating derivatives manually for a neural network would be extremely difficult.

A neural network can contain:

* Thousands of parameters
* Millions of parameters
* Billions of parameters

TensorFlow therefore provides automatic differentiation.

The `tf.GradientTape()` mechanism records the operations performed during the forward pass.

It can then use those recorded operations to calculate the gradients of the loss with respect to the model's trainable parameters.

Conceptually:

```text
Input
  ↓
Model Operations
  ↓
Prediction
  ↓
Loss
  ↓
GradientTape
  ↓
Gradients
```

This process is closely related to **backpropagation**.

---

# 8. Backpropagation

**Backpropagation** is the process used to calculate how the loss changes with respect to the parameters throughout the neural network.

Consider a simple network:

```text
Input
  ↓
Layer 1
  ↓
Layer 2
  ↓
Output
  ↓
Loss
```

The gradient information is propagated backward:

```text
Input
  ↓
Layer 1
  ↓
Layer 2
  ↓
Output
  ↓
Loss
  ↑
  │
Gradients
  ↑
  │
Layer 2
  ↑
  │
Layer 1
```

The result is a gradient for each trainable parameter.

These gradients tell the optimizer how the parameters should be changed.

---

# 9. Step 4 — Update the Weights

After calculating the gradients, the model's weights need to be updated.

This is where the **optimizer** is used.

Common optimization algorithms include:

* Stochastic Gradient Descent (SGD)
* Adam
* RMSprop
* Adagrad

The basic idea is:

```text
Current Weights
      +
Gradient Information
      ↓
Optimizer
      ↓
New Weights
```

For basic gradient descent, the update can be represented as:

$$
w_{new} = w_{old} - \eta \frac{\partial L}{\partial w}
$$

where:

* \(w_{old}\) = current weight
* \(w_{new}\) = updated weight
* \(\eta\) = learning rate
* \(\frac{\partial L}{\partial w}\) = gradient

The optimizer determines how the gradients are used to update the parameters.

---

# 10. Learning Rate

The **learning rate** controls the size of the weight updates.

A large learning rate can cause the model to make large changes:

```text
Current Weight
      ↓
   Large Step
      ↓
New Weight
```

A small learning rate results in smaller changes:

```text
Current Weight
      ↓
   Small Step
      ↓
New Weight
```

If the learning rate is too large, training may become unstable.

If it is too small, training may become very slow.

Therefore, the learning rate is an important hyperparameter when training neural networks.

---

# 11. Step 5 — Repeat for Multiple Batches

Training data is normally divided into smaller groups called **batches**.

For example, suppose we have 10,000 training examples.

Instead of processing all 10,000 examples at once, we might use:

```text
Batch size = 32
```

The training process becomes:

```text
Dataset
   ↓
Batch 1
   ↓
Forward Pass
   ↓
Loss
   ↓
Gradients
   ↓
Weight Update

Batch 2
   ↓
Forward Pass
   ↓
Loss
   ↓
Gradients
   ↓
Weight Update

Batch 3
   ↓
...
```

This continues until all training examples have been processed.

---

# 12. Epochs

An **epoch** represents one complete pass through the training dataset.

For example:

```text
Training Dataset
      ↓
Batch 1
Batch 2
Batch 3
...
Batch N
      ↓
One Epoch Complete
```

If we train for 10 epochs:

```text
Epoch 1
   ↓
Epoch 2
   ↓
Epoch 3
   ↓
...
Epoch 10
```

The model repeatedly sees the training data and updates its parameters.

---

# 13. Complete Custom Training Loop

Putting everything together:

```text
                 Training Dataset
                        ↓
                   Split into
                     Batches
                        ↓
                  ┌───────────┐
                  │  Batch    │
                  └─────┬─────┘
                        ↓
                 Forward Pass
                        ↓
                   Prediction
                        ↓
                  Compute Loss
                        ↓
                Compute Gradients
                        ↓
                 Update Weights
                        ↓
                  Next Batch
                        ↓
              All Batches Finished?
                  /           \
                No             Yes
                ↓                ↓
           Next Batch       Next Epoch
                                 ↓
                         More Epochs?
                           /       \
                         Yes        No
                          ↓          ↓
                    Next Epoch    Finished
```

This is the fundamental structure of a custom training loop.

---

# 14. The Role of `tf.GradientTape()`

One of the most important TensorFlow components in custom training loops is:

```text
tf.GradientTape()
```

Its primary purpose is to enable **automatic differentiation**.

Conceptually:

```text
GradientTape
     │
     ├── Records operations
     │
     ├── Forward pass
     │
     ├── Loss calculation
     │
     └── Calculates gradients
```

The resulting gradients can then be passed to the optimizer.

Therefore:

```text
tf.GradientTape()
        ↓
Gradients
        ↓
Optimizer
        ↓
Updated Weights
```

There are two main approaches to training a TensorFlow model.

The high-level approach:

```text

  ↓

```

Advantages:

* Simple
* Concise
* Easy to use
* Good for standard training workflows
* Integrates well with callbacks and validation

---

## Custom Training Loop

The lower-level approach:

```text
Data
 ↓

```

Advantages:

* More control

The trade-off is that you have to manage more of the training process yourself.

# 16. Multiple Loss Functions

One important reason for using a custom training loop is the need to handle **multiple losses**.

Suppose a model performs two tasks simultaneously:

```text
                  Input
                    ↓
                  Model
                 /     \
                ↓       ↓
          Classification  Regression
                ↓       ↓
             Loss 1    Loss 2
```

The total loss could be:

$$
L_{total}=L_1+L_2
$$

Or the losses could have different weights:

$$
L_{total}=\alpha L_1+\beta L_2
$$

A custom loop allows you to explicitly control how these losses are calculated and combined.

---

# 17. Multiple Models

Custom training loops can also be useful when multiple models interact.

For example:

```text
              Input
                ↓
        ┌───────┴───────┐
        ↓               ↓
     Model A          Model B
        ↓               ↓
      Loss A          Loss B
        └───────┬───────┘
                ↓
           Total Loss
                ↓
             Gradients
                ↓
        Update Parameters
```

This type of setup can be difficult to express using a simple standard training workflow.

# 18. Custom Gradient Processing

Conceptually:

```text

 ↓
Optimizer

```

For example, a specialized training algorithm might:

* Clip gradients
* Scale gradients
* Ignore certain gradients
* Apply different rules to different parameters
* Combine gradients from different sources

This gives the developer much more control over optimization.

One example of custom gradient handling is **gradient clipping**.

During training, gradients can sometimes become extremely large.

This can cause unstable training.

Conceptually:

```text
Very Large Gradient
        ↓

        ↓
Controlled Gradient
        ↓
Optimizer
        ↓
```

Gradient clipping is particularly relevant to some sequence models and deep networks.

# 21. Relationship Between the Main Concepts

It is useful to distinguish the different components:

### Forward Pass

> Uses the current weights to generate a prediction.

```text
Input → Model → Prediction
```

### Loss Function

> Measures how wrong the prediction is.

```text

```

### Gradient

> Describes how the loss changes with respect to the model's parameters.

```text
Loss → Gradients
```

### Backpropagation

> Efficiently calculates gradients throughout the network.

```text
Loss → Backward calculation → Gradients
```

### Optimizer

> Uses the gradients to determine how parameters should be updated.

```text
Gradients → Optimizer → New Weights
```

### Epoch

> One complete pass through the training dataset.

### Batch

> A subset of the training data processed in one training step.

The central idea is:

In TensorFlow, `tf.GradientTape()` is a key mechanism for implementing this process because it provides automatic differentiation, allowing the training loop to calculate the gradients needed for backpropagation and optimization.


````python
# Step 1: Import dependencies

import tensorflow as tf
import numpy as np

# Set random seed for reproducibility
tf.random.set_seed(42)
np.random.seed(42)
````

````python
# Step 2: Define a simple model: y = Wx + b

class LinearModel(tf.keras.Model):
    def __init__(self):
        super().__init__()
        self.dense = tf.keras.layers.Dense(1, activation = None)
    
    def call(self, inputs):
        return self.dense(inputs)
    
# Instantiate model
model = LinearModel()
````

````python
# Step 3: Prepare Data

# Generate synthetic data: y = 3x + 2 + noise

X_train = np.random.rand(100, 1).astype(np.float32)
y_train = 3 * X_train + 2 + 0.1 * np.random.randn(100, 1).astype(np.float32)

# Convert to tensorflow dataset
dataset = tf.data.Dataset.from_tensor_slices((X_train, y_train)).batch(10) # Batch size 10
````

````python
# Step 4: Define loss function and optimizer

# Mean square error (MSE) Loss
loss_fn = tf.keras.losses.MeanSquareError()

# Stochastic Gradient Descent (SGD) Optimizer
optimizer = tf.keras.optimizers.SGD(learning_rate = 0.1)
````

````python
# Step 5: Implementing the training loop

# Number of training epochs
epochs = 100

# Training loop

for epoch in range(epochs):
    for x_batch, y_batch in dataset:
        with tf.GradientTape() as tape:
            y_pred = model(x_batch) # Forward pass
            loss = loss_fn(y_batch, y_pred) # Compute loss

        # Compute gradients
        gradients = tape.gradients(loss, model.trainable_variables)

        # Apply gradients(Update weights)
        optimizer.apply_gradients(zip(gradients, model.trainable_variables))

    # Print progress every 10 epochs
    if epoch % 10 == 0:
        print(f"Epoch {epoch}, Loss: {loss.numpy():.4f}")
````

# Training Loop

- Start -> Initialize model <For each epoch> -> Calculate loss for input sample -> Calculate gradients of loss using gradient tape -> Update model trainable weights using gradients and learning rate. 

# Explanation of key steps
1. tf.GradientTape()
- TensorFlow´s automatic differentiation system;
- Records operations for computing gradients

2. loss_fn(y_batch, y_pred):
- Computes how far predictions are from actual values using MSE;

3. tape.gradient(loss, model.trainable_variables):
- Computes gradients of the loss with respect to model parameters;

4. optimizer.apply_gradients()
- Uses stochastic Gradient Descent(SGD) to update weight

# When to use custom training loops ?

- Need full control over training behavior;
- Implement complex loss functions or training schedules;
- Fine-tune gradients(e.g: gradient clipping);
- Work with multiple models or datasets.
