### TensorFlow & Keras Training Questions and Answers  

#### 1. Which functions are part of the built-in solution for training a model in TensorFlow and Keras?  
✅ `model.compile()` and `model.fit()`  
- `model.compile()` sets up the loss function, optimizer, and metrics.  
- `model.fit()` runs the training loop using the compiled settings.  

#### 2. Regarding the loss function, does adding the gradient to the weight (instead of subtracting) still work?  
✅ **False**  
- The gradient points in the direction of increasing loss, so we need to subtract it to minimize the loss. Adding it would increase the loss instead.  

#### 3. Which statements about the learning rate are true?  
✅ **Correct statements:**  
- "With a smaller learning rate (0.0001 as opposed to 0.1), the update to the weight is smaller, and will take more iterations to minimize the loss."  
- "It defines how big a step to take in the given direction during the update step."  

❌ **Incorrect statements:**  
- "It defines the direction in which we should jump while finding the minimum loss value." → The direction is determined by the gradient, not the learning rate.  
- "A larger learning rate (0.1 as opposed to 0.0001) leads to a bigger update and reaches the minimum loss with fewer iterations, so a larger learning rate is better." → Not always true; too large a learning rate can cause instability and prevent convergence.  

#### 4. Correct order of steps for a custom training loop:  
✅ **2, 4, 5, 3, 1**  
1. **(2)** Calculate loss  
2. **(4)** Calculate logits  
3. **(5)** Apply gradients on model using optimizer  
4. **(3)** Calculate gradients of loss with respect to the model trainable weights  
5. **(1)** Accumulate accuracy metric
