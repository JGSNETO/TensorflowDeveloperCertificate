- You want to create a tensor object that is a 2 by 3 matrix containing all -1 values. You also want to be able to modify the values inside the tensor in the future. Which of the following lines of code should you use? Check all that are true.

1. tf.Variable([-1, -1, -1, -1, -1, -1], tf.int32, shape=[2,3])

2. tf.constant([-1, -1, -1, -1, -1, -1], shape=[2, 3])
❌ Incorrect because tf.constant creates an immutable tensor, which contradicts the requirement of being able to modify the values.

3. tf.Variable([[-1, -1, -1], [-1, -1, -1]], dtype=tf.int32)
✅ Correct because tf.Variable is used with a properly nested list to create a 2×3 tensor and is modifiable.

4. tf.Variable([[-1, -1, -1], [-1, -1, -1]], shape=[2, 3])

````python
# Define a constant tensor with values [[5,7], [2,1]]
a = tf.constant([[5,7], [2,1]])

# Add 2 to each element in the tensor
b = tf.add(a,2)   # Element-wise addition: [[7,9], [4,3]]

# Square each element in the tensor
c = b ** 2        # Element-wise square: [[49,81], [16,9]]

# Compute the sum of all elements in the tensor
d = tf.reduce_sum(c) # Sum: 49 + 81 + 16 + 9 = 155
````
