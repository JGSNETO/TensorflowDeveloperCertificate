# What is a tensor

- In machine learning, particularly in TensorFlow, a tensor is the fundamental data structure. It is a multi-dimensional array similar to NumPy arrays but optimized for GPU acceleration

- A tensor has:
1. Rank(Order): The number of dimensions(e.g: scalar = 0D, vector = 1D, matrix = 2D, etc.)
2. Shape: The size of each dimension
3. Data type(dtype): The type of values stored(e.g: float32, int32)



## Creating tensors in tensorflow

- To create tensors in tensforlow, you typically use:
1. tf.constant(): Values cannot change 
2. tf.Variable(): Values can change
- You can also create tensors from NumPy arrays

```python
import tensorflow as tf

# Creating tensors
tensor1 = tf.constant([1, 2, 3])  # 1D tensor (vector)
tensor2 = tf.constant([[1, 2], [3, 4]])  # 2D tensor (matrix)

# Creating a variable tensor (mutable)
var_tensor = tf.Variable([[5.0, 6.0], [7.0, 8.0]])
```

```python
import numpy as np

np_array = np.array([[1, 2, 3], [4, 5, 6]])
tensor3 = tf.convert_to_tensor(np_array, dtype=tf.float32)
```

## Mathematical Operations on Tensor

- Tensor supports many operations similar to NumPy

```python
# Basic math operations
a = tf.constant([[1, 2], [3, 4]])
b = tf.constant([[5, 6], [7, 8]])

# Element-wise addition
c = a + b  # or tf.add(a, b)

# Element-wise multiplication
d = a * b  # or tf.multiply(a, b)

# Matrix multiplication (dot product)
e = tf.matmul(a, b)

print("Addition:\n", c.numpy())
print("Element-wise Multiplication:\n", d.numpy())
print("Matrix Multiplication:\n", e.numpy())
```

```python
# Reduction operations
tensor = tf.constant([[1, 2], [3, 4]])

# Sum of all elements
sum_all = tf.reduce_sum(tensor)

# Mean of elements
mean_val = tf.reduce_mean(tensor)

# Maximum value
max_val = tf.reduce_max(tensor)

print("Sum:", sum_all.numpy())
print("Mean:", mean_val.numpy())
print("Max:", max_val.numpy())
```

```python
# Reshaping and Transposing
tensor = tf.constant([[1, 2, 3], [4, 5, 6]])

# Reshape to a different shape
reshaped = tf.reshape(tensor, [3, 2])

# Transpose the matrix
transposed = tf.transpose(tensor)

print("Reshaped:\n", reshaped.numpy())
print("Transposed:\n", transposed.numpy())
```
