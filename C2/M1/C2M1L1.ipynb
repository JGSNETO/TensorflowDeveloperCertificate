# What is a Tensor?

In **machine learning**, and especially in **TensorFlow**, a **tensor** is the fundamental data structure used to represent all types of data. Every piece of information processed by a TensorFlow model—whether it is an image, text, audio, numerical data, model weights, predictions, or gradients—is stored as a tensor.

You can think of a tensor as a **multi-dimensional array**, similar to a NumPy array, but with additional capabilities that make it suitable for deep learning:

- Optimized for **CPU, GPU, and TPU** computation.
- Can efficiently perform parallel mathematical operations.
- Integrates with TensorFlow's automatic differentiation (`GradientTape`) for training neural networks.
- Supports distributed computing across multiple devices.

---

# Tensor Properties

Every tensor has three main properties:

## 1. Rank (Order)

The **rank** of a tensor is the **number of dimensions (axes)** it has.

It does **not** represent the size of the tensor; it only represents how many dimensions exist.

| Tensor Type | Rank | Example |
|-------------|------|---------|
| Scalar | 0 | `7` |
| Vector | 1 | `[1,2,3]` |
| Matrix | 2 | `[[1,2],[3,4]]` |
| 3D Tensor | 3 | Stack of matrices |
| 4D Tensor | 4 | Batch of images |
| 5D Tensor | 5 | Batch of videos |

### Examples

#### Rank 0 (Scalar)

A single value.

```python
tf.constant(5)
```

Shape:

```text
()
```

---

#### Rank 1 (Vector)

A list of values.

```python
tf.constant([1,2,3,4])
```

Shape:

```text
(4,)
```

---

#### Rank 2 (Matrix)

Rows and columns.

```python
tf.constant([
    [1,2],
    [3,4]
])
```

Shape:

```text
(2,2)
```

---

#### Rank 3

Multiple matrices stacked together.

```python
tf.constant([
    [[1,2],[3,4]],
    [[5,6],[7,8]]
])
```

Shape:

```text
(2,2,2)
```

---

# 2. Shape

The **shape** describes the **size of each dimension** of the tensor.

For example,

```python
tf.constant([
    [1,2,3],
    [4,5,6]
])
```

has the shape

```text
(2,3)
```

meaning:

- 2 rows
- 3 columns

Another example:

```python
image_batch = tf.zeros((32, 224, 224, 3))
```

Shape:

```text
(32, 224, 224, 3)
```

This represents:

- **32** images (batch size)
- **224** pixels height
- **224** pixels width
- **3** color channels (RGB)

This is one of the most common tensor shapes in computer vision.

---

# 3. Data Type (dtype)

The **dtype** specifies the type of values stored inside the tensor.

Common TensorFlow data types include:

| Data Type | Description |
|------------|-------------|
| `tf.float32` | 32-bit floating point (most common for neural networks) |
| `tf.float64` | 64-bit floating point |
| `tf.int32` | 32-bit integer |
| `tf.int64` | 64-bit integer |
| `tf.bool` | Boolean values (`True`/`False`) |
| `tf.string` | Text strings |
| `tf.uint8` | Unsigned 8-bit integer (common for raw images) |

Example:

```python
tf.constant([1,2,3], dtype=tf.float32)
```

Output:

```text
dtype=float32
```

# Why TensorFlow Uses Tensors

Unlike regular Python lists, tensors are specifically designed for high-performance numerical computation.

Advantages include:

- Efficient memory management.
- Parallel execution on GPUs and TPUs.
- Fast matrix operations.
- Automatic differentiation for backpropagation.
- Easy distribution across multiple devices.
- Optimized computational graphs for deep learning.

# Tensors in Machine Learning

Almost everything inside a neural network is represented as tensors.

| Component | Typical Tensor Shape |
|-----------|----------------------|
| Single feature vector | `(features,)` |
| Batch of samples | `(batch_size, features)` |
| Image | `(height, width, channels)` |
| Batch of images | `(batch_size, height, width, channels)` |
| Sequence of words | `(sequence_length,)` |
| Batch of sequences | `(batch_size, sequence_length)` |
| Video | `(frames, height, width, channels)` |
| Batch of videos | `(batch_size, frames, height, width, channels)` |

Examples:

A batch of **64 RGB images** of size **128×128**:

```text
(64, 128, 128, 3)
```

A batch of **32 sentences**, each with **100 tokens**:

```text
(32, 100)
```

# Relationship Between Rank and Shape

These concepts are closely related but describe different properties:

| Shape | Rank | Meaning |
|--------|------|---------|
| `()` | 0 | Scalar |
| `(5,)` | 1 | Vector with 5 elements |
| `(3,4)` | 2 | Matrix with 3 rows and 4 columns |
| `(32,224,224,3)` | 4 | Batch of RGB images |
| `(8,30,224,224,3)` | 5 | Batch of videos |

Notice that:

- **Rank** = number of dimensions.
- **Shape** = size along each dimension.

# Summary

A **tensor** is the core data structure in TensorFlow and the foundation of all machine learning computations.

Every tensor is characterized by three key properties:

- **Rank** – the number of dimensions (axes).
- **Shape** – the size of each dimension.
- **Data Type (`dtype`)** – the type of values stored in the tensor.

Because tensors support efficient operations on CPUs, GPUs, and TPUs, they enable TensorFlow to train and run deep learning models at high speed. Understanding tensors is essential before learning about neural networks, as every input, weight, activation, gradient, and prediction in TensorFlow is represented as a tensor.

## Creating tensors in tensorflow

- To create tensors in tensforlow, you typically use:
1. tf.constant(): Values cannot change 
2. tf.Variable(): Values can change
- You can also create tensors from NumPy arrays

````python
import tensorflow as tf

# Creating tensors
tensor1 = tf.constant([1, 2, 3])  # 1D tensor (vector)
tensor2 = tf.constant([[1, 2], [3, 4]])  # 2D tensor (matrix)

# Creating a variable tensor (mutable)
var_tensor = tf.Variable([[5.0, 6.0], [7.0, 8.0]])
````

````python
import numpy as np

np_array = np.array([[1, 2, 3], [4, 5, 6]])
tensor3 = tf.convert_to_tensor(np_array, dtype=tf.float32)
````

## Mathematical Operations on Tensor

- Tensor supports many operations similar to NumPy

````python
# Basic math operations
a = tf.constant([[1, 2], [3, 4]])
b = tf.constant([[5, 6], [7, 8]])

# Element-wise addition
c = a + b  # or tf.add(a, b)

# Element-wise multiplication
d = a * b  # or tf.multiply(a, b)

# Matrix multiplication (dot product)
e = tf.matmul(a, b)

print("Addition:\n", c.numpy())
print("Element-wise Multiplication:\n", d.numpy())
print("Matrix Multiplication:\n", e.numpy())
````

````python
# Reduction operations
tensor = tf.constant([[1, 2], [3, 4]])

# Sum of all elements
sum_all = tf.reduce_sum(tensor)

# Mean of elements
mean_val = tf.reduce_mean(tensor)

# Maximum value
max_val = tf.reduce_max(tensor)

print("Sum:", sum_all.numpy())
print("Mean:", mean_val.numpy())
print("Max:", max_val.numpy())
````

````python
# Reshaping and Transposing
tensor = tf.constant([[1, 2, 3], [4, 5, 6]])

# Reshape to a different shape
reshaped = tf.reshape(tensor, [3, 2])

# Transpose the matrix
transposed = tf.transpose(tensor)

print("Reshaped:\n", reshaped.numpy())
print("Transposed:\n", transposed.numpy())
````
