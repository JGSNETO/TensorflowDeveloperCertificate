## **Question 1**  
Which of the following statements is **false** about the Graph approach?

- ✅ **Faster compilation**  
- ❌ Easier debugging  
- ❌ Parallelism  
- ❌ Portability  

### **Explanation:**  
The Graph approach in TensorFlow offers benefits like **easier debugging**, **parallelism**, and **portability**, but **it does not lead to faster compilation**. Building and optimizing the computation graph takes time before execution.

## **Question 2**  
Which of the following statements is **true** for `tf.cond`?

- ✅ **Graph execution does not support if/else statements. To replicate that effect you use `tf.cond`.**  
- ❌ `tf.cond` is an alternative to using if/else statements in Graphs, as its execution is much faster than if/else statements.

### **Explanation:**  
Since TensorFlow’s computation graphs **do not execute Python's native if/else**, `tf.cond` is used to **conditionally execute branches** during graph execution.
