# TPU Strategy in Tensorflow

- tf.distribute.TPUStrategy is a tensorflow distribution strategy designed for training models on tensor processing units (TPUs). TPUs are specialized hardware accelerators optimized for deep learning workloads, providing significant speedups over CPUs and GPUs in certain scenarios.

## How TPU strategy works
1. Model Replication Across TPU Cores
- The model is replicated across multiple TPU Cores. Each TPU core runs the same model independently. 
2. Data Parallelism
- Input data is sharded(split across TPU cores).
- Each core processes its own subset of data. 
3. Synchronous Training with Cross-Replica Summarization
- Each TPU core computes gradients for its data batch.
- Gradients from all TPU cores are aggregated and averaged. 
- Model parameters are updated synchronously to maintain consistency across all TPU cores.


```python
import tensorflow as tf

# Detect and initialize TPU
resolver = tf.distribute.cluster_resolver.TPUClusterResolver.connect()
strategy = tf.distribute.TPUStrategy(resolver)

# Define and compile the model inside strategy scope
with strategy.scope():
    model = tf.keras.Sequential([
        tf.keras.layers.Dense(128, activation='relu', input_shape=(784,)),
        tf.keras.layers.Dense(10, activation='softmax')
    ])

    model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])

# Load and preprocess data
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
x_train, x_test = x_train / 255.0, x_test / 255.0

# Train the model on TPU
model.fit(x_train, y_train, batch_size=128, epochs=5)
```

# When to use TPU Strategy
- Training on TPUs, such as Google Cloud TPUs or TPU VMs.
- Large-scale models requiring fast matrix multiplications.
- High-speed training with very large datasets(e.g: ImageNet, BERT)
- Batch sizes that are large(since TPUs perform best with larger batch sizes).

# Alternatices to TPU Strategy
- Mirrored Strategy: Multi-GPU training.
- MultiWorkerMirroredStrategy: Multi-node, multi-gpu training.
- ParameterServerStrategy: Asynchronous distributed training.
