# Mirrored Stategy

- In the context of TensorFlow and machine learning, Mirrored strategy(tf.distribute.MirroredStrategy) is a data parallelism strategy used for distributed training across multiple GPUs on a single machine. It enables synchronous training by replicating the model across multiple devices and keeping them in sync.

## How it works
1. Model replication: The model, including its variables, is copied(mirrored) accross all available GPUs. 
2. Data parallelism: Each GPU gets a different mini-batch of the input data. 
3. Synchronous Training: Each GPU computes forward and backward passes independently, and gradients are computed separately. 
4. Gradient Aggregation: The gradients from all GPUs are averaged(all-redice operations) before updating the model parameters, ensuring all GPUs stay synchrozined

```python
import tensorflow as tf

# Define MirroredStrategy
strategy = tf.distribute.MirroredStrategy()

print(f"Number of devices: {strategy.num_replicas_in_sync}")

# Use strategy.scope() to create and compile the model
with strategy.scope():
    model = tf.keras.Sequential([
        tf.keras.layers.Flatten(input_shape = (28, 28)),
        tf.keras.layers.Dense(128, activation='relu'),
        tf.keras.layers.Dense(10, activation='softmax')
    ])
    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# Load dataset
(train_images, train_labels), _ = tf.keras.datasets.mnist.load_data()
train_images = train_images / 255.0

# Train model
model.fit(train_images, train_labels, epochs=5, batch_size=64)
```

# When to use Mirrored Stategu
- When training large models that require multiples GPUs.
- When training on a single machine with multipe GPUs.
- When you need synchronous training for consistency.

# Alternatives:
- tf.distribute.MultiWorkerMirroredStrategy: For multiple machines.
- tf.distribute.TPUStrategy: For TPU-based training. 
- tf.distribute.ParameterServerStrategy: For large-scale distributed training
